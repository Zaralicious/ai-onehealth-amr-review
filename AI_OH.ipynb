```python
import numpy as np

def simplify_ai(text):
    text = str(text).lower()
    if any(k in text for k in ['xgboost', 'gradient boosting', 'gbm', 'catboost', 'lightgbm']):
        return 'Gradient Boosting Methods'
    elif any(k in text for k in ['random forest', 'decision tree', 'tree-based']):
        return 'Tree-Based Models'
    elif any(k in text for k in ['support vector', 'svm']):
        return 'Support Vector Machines'
    elif any(k in text for k in ['neural network', 'cnn', 'rnn', 'lstm', 'deep learning', 'u-net', 'resnet']):
        return 'Neural / Deep Learning Models'
    elif any(k in text for k in ['machine learning', 'ml model', 'supervised', 'unsupervised']):
        return 'General Machine Learning'
    elif any(k in text for k in ['nlp', 'natural language', 'text mining', 'language model', 'chatgpt']):
        return 'Natural Language Processing'
    elif any(k in text for k in ['regression', 'logistic regression', 'glm']):
        return 'Regression Models'
    elif any(k in text for k in ['clustering', 'k-means', 'hierarchical']):
        return 'Clustering / Unsupervised Learning'
    elif any(k in text for k in ['ensemble', 'metaheuristic', 'hybrid']):
        return 'Ensemble / Hybrid Models'
    elif any(k in text for k in ['computer vision', 'image', 'segmentation', 'classification']):
        return 'Computer Vision'
    elif any(k in text for k in ['reinforcement']):
        return 'Reinforcement Learning'
    elif any(k in text for k in ['forecast', 'sarima', 'time series']):
        return 'Time Series / Forecasting'
    else:
        return 'Other / Mixed'

df['AI_Category'] = df['AI_Technique_Used'].apply(simplify_ai)
```

```python
# STEP 5: CLUSTERING PURPOSE OF AI APPLICATIONS

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Use your cleaned text column
text_data = df['Purpose_of_AI_Application'].dropna().astype(str)

# Convert to lowercase and vectorize
vectorizer = TfidfVectorizer(stop_words='english', max_df=0.9, min_df=2)
X = vectorizer.fit_transform(text_data)

# Apply K-Means (try 5 clusters first — we can adjust later)
kmeans = KMeans(n_clusters=5, random_state=42)
kmeans.fit(X)

# Add cluster labels to the dataframe
df['Cluster'] = kmeans.labels_

# Display number of items in each cluster
print(df['Cluster'].value_counts())

# View sample purposes per cluster
for i in range(5):
    print(f"\n🔹 Cluster {i}:")
    samples = df[df['Cluster'] == i]['Purpose_of_AI_Application'].sample(3, random_state=42)
    for text in samples:
        print("-", text[:200], "...")
```

```python
# STEP 7: VISUALIZING CLUSTERS

from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

# Reduce TF-IDF data to 2D using PCA
pca = PCA(n_components=2, random_state=42)
reduced_data = pca.fit_transform(X.toarray())

# Create a DataFrame for plotting
cluster_df = pd.DataFrame(reduced_data, columns=['PCA1', 'PCA2'])
cluster_df['Cluster'] = kmeans.labels_

# Plot the clusters
plt.figure(figsize=(10, 7))
sns.scatterplot(
    x='PCA1', 
    y='PCA2', 
    hue='Cluster', 
    data=cluster_df, 
    palette='tab10', 
    s=60, 
    alpha=0.8
)
plt.title('AI–One Health Publications Cluster Visualization', fontsize=14)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
plt.savefig('AI–One Health Publications Cluster Visualization.png', dpi=300, bbox_inches='tight')
```

```python
import pandas as pd

def extract_ai_technique(text):
    text = str(text).lower()
    techniques = []
    if 'xgboost' in text or 'gradient boosting' in text or 'gbm' in text:
        techniques.append('XGBoost / Gradient Boosting')
    if 'svm' in text or 'support vector' in text:
        techniques.append('SVM')
    if 'random forest' in text or 'rf' in text:
        techniques.append('Random Forest')
    if 'logistic regression' in text or 'regression' in text:
        techniques.append('Regression')
    if 'neural network' in text or 'cnn' in text or 'deep learning' in text or 'rnn' in text or 'lstm' in text or 'ann' in text:
        techniques.append('Neural Network / Deep Learning')
    if 'decision tree' in text or 'tree' in text:
        techniques.append('Decision Tree')
    if 'k-means' in text or 'clustering' in text:
        techniques.append('Clustering')
    if 'nlp' in text or 'natural language' in text or 'text mining' in text:
        techniques.append('NLP / Text Mining')
    if 'pca' in text:
        techniques.append('PCA')
    if 'ensemble' in text:
        techniques.append('Ensemble Model')
    if 'bayes' in text or 'naive bayes' in text:
        techniques.append('Naive Bayes')
    if not techniques:
        techniques.append('Other / Unspecified')
    return ', '.join(set(techniques))

df['Simplified_Technique'] = df['AI_Technique_Used'].apply(extract_ai_technique)

df['Simplified_Technique'].value_counts().head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,8))
sns.countplot(y='Simplified_Technique', data=df, order=df['Simplified_Technique'].value_counts().index)
plt.title('Most Common AI Techniques Used in One Health Research')
plt.xlabel('Number of Publications')
plt.ylabel('AI Technique')
plt.tight_layout()
plt.show()
plt.savefig('Most Common AI Techniques Used in One Health Research.png', dpi=300, bbox_inches='tight')
```
